```python
#Basic mean (Easy)
#Compute the mean of the eval scores array. Then compute the mean of only the scores above 80. Compare the two.
scores = np.array([72, 85, 90, 61, 78, 95, 88, 73, 82, 79])

# 1a. Mean of all scores
a = np.mean(scores)
print(a)

# 1b. Mean of scores above 80 only
b = np.mean(scores[scores > 80])
print(b)

# 1c. How much higher is the filtered mean vs overall mean?
print(round((b-a), 2))
```

```python
# Axis-wise mean on a 2D array (Easy)
#You have a matrix where each row is a model and each column is an eval run score. Compute the mean score per model (row mean) and the mean score per eval run (column mean).
model_scores = np.array([
    [82, 79, 85, 88],   # GPT-4
    [74, 80, 76, 83],   # Claude
    [91, 88, 94, 90],   # Gemini
])

# 2a. Mean score per model (one value per row)
row_means = model_scores.mean(axis=1)
print(f"Mean per eval run: {row_means}")

# 2b. Mean score per eval run (one value per column)
col_means = model_scores.mean(axis=0)
print(f"Mean per eval run: {col_means}")

# 2c. Which model performed best on average?
models = ['GPT-4', 'Claude', 'Gemini']
best = models[np.argmax(row_means)]
print(f"Best model: {best}")
```

```python
#Weighted mean (Medium)
#In an LLM eval, not all test cases are equally important. Compute a weighted mean score where harder questions count more.
scores  = np.array([90, 75, 60, 85, 70])
weights = np.array([1,   2,  3,  2,  1])   # harder questions weighted higher

# Weighted mean = sum(score * weight) / sum(weights)
# Expected: ~73.33
result = np.sum(scores * weights) / np.sum(weights)
print(f"weighted mean: {result:.2f}")

#Method 2 - np.average()
weighted_mean = np.average(scores, weights = weights)
print(f"weighted mean using np.average function: {weighted_mean:.2f}")
```

```python
#Rolling mean (Mean over a window) (Medium)
#Compute a 3-point rolling mean of the latency array — each value is the mean of itself and the two values before it. This smooths out spikes.
import numpy as np
import pandas as pd

latencies = np.array([420, 890, 5100, 760, 210, 1100, 340, 280, 6200, 650])

s = pd.Series(latencies)
print(s.rolling(window = 3).mean().round(2).values) #This will keep NaN's as it is in the result
#Calculates the mean using 1 point for the first row, 2 points for the second row
print(s.rolling(window=3, min_periods=1).mean().round(2).values) #min_periods=1 - This will not give NaN's from the array to be used for machine learning model or database which cannot accept missing data
# Rolling mean window=3:
# index 0: can't compute (not enough history) → NaN
# index 1: can't compute → NaN
# index 2: mean(420, 890, 5100) = 2136.67
# index 3: mean(890, 5100, 760) = 2250.0
# ... and so on

```

```python
#Compute variance and std dev (Easy)
#Compute the variance and standard deviation of the eval scores. Then interpret what they tell you about model consistency.
import numpy as np

scores = np.array([72, 85, 90, 61, 78, 95, 88, 73, 82, 79])

# 5a. Population variance and std dev (ddof=0)
pop_var = np.var(scores, ddof=0)
pop_std = np.std(scores, ddof=0)
print(f"Population variance is {pop_var:.2f} and population standard deviation is {pop_std:.2f}")

# 5b. Sample variance and std dev (ddof=1)
sam_var = np.var(scores, ddof=1)
sam_std = np.std(scores, ddof=1)
print(f"Sample variance is {sam_var:.2f} and sample standard deviation is {sam_std:.2f}")

# 5c. What does a std dev of ~10 mean in plain English?
regular_mean = scores.mean()
print(f"Mean is {regular_mean:.2f}")
print(f"Most scores fall between {regular_mean - sam_std:.1f} and {regular_mean + sam_std:.1f}")
```

```python
#Compare variance across models (Medium)
# Two models are evaluated 6 times each. Compare their consistency using std dev.
import numpy as np

model_a = np.array([82, 84, 81, 83, 82, 85])   # consistent?
model_b = np.array([70, 95, 78, 91, 65, 88])   # consistent?

# 6a. Compute mean and std dev for both
mean_a, std_a = model_a.mean(), np.std(model_a, ddof=1)
mean_b, std_b = model_b.mean(), np.std(model_b, ddof=1)
print(f"Mean and std for Model A is {mean_a:.2f} and {std_a:.2f}")
print(f"Mean and std for Model B is {mean_b:.2f} and {std_b:.2f}")

# 6b. Which model is more consistent?
consistent_model = "Model A" if std_a < std_b else "Model B"
print(f"More consistent model is: {consistent_model}")

# 6c. Which would you prefer to deploy and why?
print("\nDeploy Model A:")
print(f"  Similar mean ({mean_a:.1f} vs {mean_b:.1f})")
print(f"  But 8x more consistent (std {std_a:.1f} vs {std_b:.1f})")
print(f"  Model B's range: [{model_b.min()}, {model_b.max()}]")
print(f"  Model A's range: [{model_a.min()}, {model_a.max()}]")
```

```python
# Normalisation using mean and std (Z-score) (Medium)

latencies = np.array([420, 890, 5100, 760, 210, 1100, 340, 280, 6200, 650])

mean = latencies.mean()
std  = latencies.std(ddof=1)

z_scores = (latencies - mean) / std

print(f"Original mean: {mean:.1f}, std: {std:.1f}")
print(f"Normalised mean: {z_scores.mean():.6f}, std: {z_scores.std(ddof=1):.6f}")
# Normalised — mean: 0.000000, std: 1.000000

print(f"\nZ-scores: {np.round(z_scores, 2)}")
# Negative = faster than average, Positive = slower than average
# Large positive values (timeout calls) stand out clearly as outliers
```

```python
# Coefficient of Variation (Hard)
# The Coefficient of Variation (CV) is std / mean * 100 — it expresses variability as a percentage of the mean. This lets you compare variability across datasets with different scales.
# Compare the CV of latencies vs costs to find out which is more variable relative to its scale.

import numpy as np

latencies = np.array([420, 890, 5100, 760, 210, 1100, 340, 280, 6200, 650])
costs     = np.array([0.013, 0.019, 0.0, 0.015, 0.005, 0.046, 0.005, 0.008, 0.0, 0.012])

# CV = (std / mean) * 100
# Which is more variable — latency or costs
std_latencies = latencies.std(ddof=1)
std_costs = costs.std(ddof=1)

cv_latencies = std_latencies / latencies.mean() * 100
cv_costs = std_costs / costs.mean() * 100

print(f"Latencies - Mean: {latencies.mean():.2f}, std: {std_latencies:.2f}, CV: {cv_latencies:.2f}")
print(f"Costs - Mean: {costs.mean():.2f}, std: {std_costs:.2f}, CV: {cv_costs:.2f}")
print(f"More Variable: {"Latency" if cv_latencies > cv_costs else "Costs"}")
```

```python
#Broadcasting (Medium)
# You have a 4x3 matrix of model evaluation scores across 3 tasks. Normalise each row so values fall between 0 and 1 (min-max normalisation per model).
#normalised = (score - row_min) / (row_max - row_min)
import numpy as np

scores = np.array([
    [60, 80, 90],    # Model A across 3 tasks
    [50, 95, 70],    # Model B
    [85, 85, 85],    # Model C
    [40, 60, 100],   # Model D
])

row_min = scores.min(axis = 1, keepdims=True) #keepdims=True is critical — without it, row_min has shape (4,) instead of (4,1) and broadcasting fails.
row_max = scores.max(axis = 1, keepdims=True)

normalized_array = (scores - row_min) / (row_max - row_min)
print(np.round(normalized_array, 2))

# Handle Model C (all same scores — division by zero → nan)
normalised = np.nan_to_num(normalized_array, nan=0.0)  #nan_to_num catches the 0/0 case where all scores are identical.
print(np.round(normalised, 2))
```

```python
#Dot product and cosine similarity (Medium)
#Compute the dot product between two embedding vectors. Then compute the cosine similarity between them. Cosine similarity = dot product of unit vectors.
vec_a = np.array([0.2, 0.8, 0.4, 0.6])   # embedding for "machine learning"
vec_b = np.array([0.3, 0.7, 0.5, 0.5])   # embedding for "deep learning"
vec_c = np.array([0.9, 0.1, 0.1, 0.2])   # embedding for "cooking recipes"

# 11a. Dot product of vec_a and vec_b
print(np.dot(vec_a, vec_b).round(4))

# Cosine similarity between vec_a and vec_b, vec_a and vec_c
def cosine_sim(x, y):
    return np.dot(x, y) / (np.linalg.norm(x) * np.linalg.norm(y))

sim_ab = cosine_sim(vec_a, vec_b)
sim_ac = cosine_sim(vec_a, vec_c)

print(f"Cosine sim a vs b: {sim_ab:.4f}")   # ~0.98 — very similar
print(f"Cosine sim a vs c: {sim_ac:.4f}")   # ~0.41 — less similar

# 11d
more_similar = 'vec_b' if sim_ab > sim_ac else 'vec_c'
print(f"More similar to vec_a: {more_similar}")   # vec_b
```

```python
# Matrix multiplication for batch similarity (Hard)
# Instead of computing cosine similarity one pair at a time, compute it for an entire batch using matrix multiplication. Given one query vector and 5 document vectors, compute all similarities at once.

np.random.seed(42)
query = np.random.randn(4)
docs  = np.random.randn(5, 4)

print(query)
print(docs)

# Step 1 — normalise query
query_norm = query / np.linalg.norm(query)

# Step 2 — normalise each doc row
doc_norms  = np.linalg.norm(docs, axis=1, keepdims=True)   # shape (5,1)
docs_norm  = docs / doc_norms                               # shape (5,4)

# Step 3 — batch cosine similarity via matrix multiply
# query_norm shape: (4,)  →  treat as (1,4)
# docs_norm.T shape: (4,5)
# result shape: (5,)
similarities = docs_norm @ query_norm   # equivalent to query_norm @ docs_norm.T

print(f"Similarities: {np.round(similarities, 4)}")
# e.g. [ 0.3801 -0.1234  0.5621  0.2198 -0.4012]

best_doc = np.argmax(similarities)
print(f"Most similar doc: index {best_doc} with score {similarities[best_doc]:.4f}")
```

```python
#Statistical summary of embedding space (Hard)
#Given a matrix of embeddings, compute a full statistical profile — mean vector (centroid), variance per dimension, and find which document is closest to the centroid.
import numpy as np

np.random.seed(7)
embeddings = np.random.randn(8, 6)   # 8 documents, 6-dim embeddings
print(embeddings)

# 14a. Centroid — mean vector across all documents (shape: 6,)
centroid = embeddings.mean(axis = 0)
print(f"Centroid: {np.round(centroid, 3)}")

# 14b. Variance per dimension (how spread out is each dimension?)
variance = embeddings.var(axis=0, ddof=1)
print(f"Variance: {np.round(variance, 3)}")

# 14c. Which document is closest to the centroid? (use Euclidean distance)
distances = np.linalg.norm(embeddings-centroid, axis = 1)
shorter_index = np.argmin(distances)
print(f"Distances to centroid: {np.round(distances, 4)}")
print(f"Document index closest to centroid is {shorter_index} and the document is: {np.round(embeddings[shorter_index],3)} ")
print(f"Closer document distance: {distances[shorter_index]:.3f}")
print("Shape:", distances.shape)

# 14d. Which dimension has the highest variance?
high_var_idx = np.argmax(variance)
print(f"High variance index is {high_var_idx} and the value is {np.round(variance[high_var_idx], 3)}")
```

```python
# COMBINED CHALLENGE (Hard).
#Build embedding_stats(embeddings) that returns a full analysis dict:

# centroid — mean vector (shape: n_dims,)
# mean_pairwise_similarity — average cosine similarity between all pairs of documents
# most_similar_pair — tuple of indices (i, j) with highest cosine similarity
# least_similar_pair — tuple of indices (i, j) with lowest cosine similarity
# dim_variance — variance per dimension
# most_variable_dim — index of most variable dimension
import numpy as np

def embedding_stats(embeddings):

  variance = embeddings.var(axis= 0, ddof=1).round(4)

  norms = np.linalg.norm(embeddings, axis=1, keepdims=True)
  normed = embeddings / norms

  sim_matrix = normed @ normed.T
  np.fill_diagonal(sim_matrix, np.nan)

   # Most and least similar pairs
  flat_idx_max = np.nanargmax(sim_matrix)
  flat_idx_min = np.nanargmin(sim_matrix)
  i_max, j_max = np.unravel_index(flat_idx_max, sim_matrix.shape)
  i_min, j_min = np.unravel_index(flat_idx_min, sim_matrix.shape)

  return {
      'centroid' : embeddings.mean(axis=0).round(4),
      'mean_pairwise_similarity': round(float(np.nanmean(sim_matrix)),4),
      'most_similar_pair':  (int(i_max), int(j_max)),
      'least_similar_pair':   (int(i_min), int(j_min)),
      'dim_variance' : variance,
      'most_variable_dim': int(np.argmax(variance))

  }

np.random.seed(0)
embeddings = np.random.randn(6, 4)   # 6 docs, 4-dim

result = embedding_stats(embeddings)
print(result)
print(result['most_similar_pair'])    # e.g. (1, 4)
print(result['mean_pairwise_similarity'])  # e.g. -0.05
```
